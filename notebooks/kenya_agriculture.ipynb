### Handling Missing Values

The missing value assessment showed that the `Value` column contains 731 missing observations. Inspection revealed that these records are marked with the flag `M`, indicating that the measurements were unavailable rather than equal to zero.

To preserve the integrity of the dataset, missing values were not imputed. Instead, records with missing `Value` entries were excluded from the cleaned dataset used for subsequent transformations and analysis.

The `Note` column contains many missing values because it stores optional annotations. These missing values do not affect the analytical variables and were therefore retained.

```python
# Filter  data

# Production
production_df = filter_by_element(clean_df, "Production")
production_df.show(3)

# Area Harvested
harvest_df = filter_by_element(clean_df, "Area harvested")
harvest_df.show(3)

# Yield
yield_df = filter_by_element(clean_df, "Yield")
yield_df.show(3)

#Stocks 

stocks_df = filter_by_element(clean_df, "Stocks")
stocks_df.show(3)

#laying 
laying_df = filter_by_element(clean_df, "Laying")
laying_df.show(3)

#Milk anaimals

milk_animals_df = filter_by_element(clean_df, "Milk Animals")
milk_animals_df.show(3)

#"Producing Animals/Slaughtered 

slaughtered_df = filter_by_element(
    clean_df,
    "Producing Animals/Slaughtered"
)
slaughtered_df.show(3)

#car"Yield/Carcass Weight

carcass_df = filter_by_element(
    clean_df,
    "Yield/Carcass Weight"
)
carcass_df.show(3)
```

```python
# Production
production_analysis_df = prepare_analysis_dataset(production_df)

# Area Harvested
harvest_analysis_df = prepare_analysis_dataset(harvest_df)

# Yield
yield_analysis_df = prepare_analysis_dataset(yield_df)

# Stocks
stocks_analysis_df = prepare_analysis_dataset(stocks_df)

# Laying
laying_analysis_df = prepare_analysis_dataset(laying_df)

# Milk Animals
milk_analysis_df = prepare_analysis_dataset(milk_animals_df)

# Producing Animals / Slaughtered
slaughtered_analysis_df = prepare_analysis_dataset(slaughtered_df)

# Yield / Carcass Weight
carcass_analysis_df = prepare_analysis_dataset(carcass_df)
```
