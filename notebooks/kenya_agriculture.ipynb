### Handling Missing Values

The missing value assessment showed that the `Value` column contains 731 missing observations. Inspection revealed that these records are marked with the flag `M`, indicating that the measurements were unavailable rather than equal to zero.

To preserve the integrity of the dataset, missing values were not imputed. Instead, records with missing `Value` entries were excluded from the cleaned dataset used for subsequent transformations and analysis.

The `Note` column contains many missing values because it stores optional annotations. These missing values do not affect the analytical variables and were therefore retained.

```python
# Standardize column names

from scripts.clean_data import standardize_column_names
clean_df = standardize_column_names(clean_df)
print(clean_df.columns)
```
